```python
<VSCode.Cell language="markdown">
# Random Arrays with NumPy
This notebook demonstrates deterministic random array generation using NumPy, sampling from distributions, visualization, persistence, and unit testing.
</VSCode.Cell>
<VSCode.Cell language="python">
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pytest

sns.set_theme(style="darkgrid")
np.set_printoptions(precision=4, suppress=True)
print("NumPy version:", np.__version__)
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Configure Notebook & Reproducibility
Use `np.random.default_rng` for reproducible randomness and set a fixed seed.
</VSCode.Cell>
<VSCode.Cell language="python">
seed = 12345
rng = np.random.default_rng(seed)
print("RNG created with seed", seed)
print("Sample scalar:", rng.integers(0, 10))
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Create Random Arrays (1D, 2D, 3D)
Examples of arrays created with `rng.random` and `np.random.rand`.
</VSCode.Cell>
<VSCode.Cell language="python">
arr1d = rng.random(8, dtype=np.float64)
arr2d = rng.random((3, 4), dtype=np.float32)
arr3d = np.random.rand(2, 3, 4).astype(np.float64)

print("1D array:", arr1d)
print("2D array shape", arr2d.shape)
print("3D array shape", arr3d.shape)
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Generate Random Integers
Use `rng.integers` and `np.random.randint` for integer samples.
</VSCode.Cell>
<VSCode.Cell language="python">
ints = rng.integers(0, 20, size=(4, 3))
ints_neg = np.random.randint(-5, 5, size=5)
print("Integer matrix:\n", ints)
print("Negative range sample:", ints_neg)
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Generate Random Floats and Ranges
Create float arrays over different ranges and dtypes.
</VSCode.Cell>
<VSCode.Cell language="python">
float_samples = rng.random(5, dtype=np.float32) * 10.0
linspace_samples = np.linspace(0, 1, 5, dtype=np.float64)
print("Scaled floats:", float_samples)
print("Linspace:", linspace_samples)
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Sample from Probability Distributions
Draw from normal, binomial, and poisson distributions.
</VSCode.Cell>
<VSCode.Cell language="python">
normal_sample = rng.normal(loc=0, scale=1, size=1000)
binomial_sample = rng.binomial(n=10, p=0.4, size=1000)
poisson_sample = rng.poisson(lam=3.5, size=1000)
print("Normal mean, std:", normal_sample.mean(), normal_sample.std())
print("Binomial unique:", np.unique(binomial_sample))
print("Poisson unique:", np.unique(poisson_sample)[:10])
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Random Permutations, Shuffling, and Sampling without Replacement
Use `rng.permutation`, `rng.shuffle`, and `rng.choice(replace=False)`.
</VSCode.Cell>
<VSCode.Cell language="python">
base = np.arange(10)
perm = rng.permutation(base)
shuffled = base.copy()
rng.shuffle(shuffled)
sampled = rng.choice(base, size=5, replace=False)
print("Permutation:", perm)
print("Shuffled:", shuffled)
print("Sampled without replacement:", sampled)
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Random Choice with Probabilities and Weights
Sample from weighted probability distributions with `rng.choice`.
</VSCode.Cell>
<VSCode.Cell language="python">
choices = ['red', 'green', 'blue']
weights = [0.1, 0.7, 0.2]
weighted = rng.choice(choices, size=10, replace=True, p=weights)
print("Weighted choices:", weighted)
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Structured and Masked Random Arrays
Create structured arrays and masked arrays for random fields.
</VSCode.Cell>
<VSCode.Cell language="python">
dtype = [('x', 'f4'), ('y', 'i4'), ('label', 'U6')]
structured = np.zeros(5, dtype=dtype)
structured['x'] = rng.random(5)
structured['y'] = rng.integers(0, 100, size=5)
structured['label'] = rng.choice(['red', 'green', 'blue'], size=5)
mask = rng.random(5) < 0.3
masked = np.ma.array(structured, mask=np.column_stack([mask, mask, mask]))
print("Structured array:\n", structured)
print("Masked array:\n", masked)
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Performance, Memory Views, and Avoiding Copies
Demonstrate views, reshape, and memory layout.
</VSCode.Cell>
<VSCode.Cell language="python">
large = rng.random((1000, 1000), dtype=np.float64)
view = large[:, :500]
copy = large[:, :500].copy()
print("view shares memory with large?", np.shares_memory(large, view))
print("copy shares memory with large?", np.shares_memory(large, copy))
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Visualize Random Data
Plot distribution histograms and scatter plots.
</VSCode.Cell>
<VSCode.Cell language="python">
plt.figure(figsize=(10, 4))
plt.hist(normal_sample, bins=30, alpha=0.7)
plt.title('Normal Distribution Sample')
plt.show()

plt.figure(figsize=(5, 5))
plt.scatter(np.arange(100), normal_sample[:100], alpha=0.6)
plt.title('First 100 Normal Samples')
plt.show()
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Save, Load, and Serialize Random Arrays
Persist arrays and verify round-trip equality.
</VSCode.Cell>
<VSCode.Cell language="python">
import tempfile
with tempfile.TemporaryDirectory() as d:
    path = f"{d}/array.npy"
    np.save(path, arr2d)
    loaded = np.load(path)
    print("Saved and loaded shapes:", loaded.shape)
    print("Arrays equal:", np.array_equal(arr2d, loaded))
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Unit Tests for Deterministic Randomness
Use `pytest` to validate deterministic array generation.
</VSCode.Cell>
<VSCode.Cell language="python">
def test_rng_deterministic():
    rng_test = np.random.default_rng(12345)
    arr = rng_test.integers(0, 10, size=5)
    assert np.array_equal(arr, np.array([3, 0, 3, 7, 9]))

print("Test function defined. Run `pytest random_arrays.ipynb` or convert to .py for automated execution.")
</VSCode.Cell>
```
